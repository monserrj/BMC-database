# Complex
<b> A complex means: "A shell structure that has been successfully assembled, with or without enzymatic activity." A complex is uniquely identified by the set of protein sequences that compose it.

Therefore a complex is not simply a list of compatible proteins. There needs to be evidence that the combination can actually assemble.This evidence can be via simulation and modelling (theoretical)

Examples of separate complexes could therefore be:

- Complex 1: PduA + PduB + PduJ + PduK + PduN
- Complex 2: PduA + PduJ + PduK
- Complex 3: PduA + PduB + PduJ + PduK + PduN + PduU + PduT
- Complex 4 (chimeric:proteins do not necessarily have to originate from the same natural BMC system.
): PduA + EutS

<b> A complex does not require every member protein to directly interact with every other member. 
<b> We may know the complex but not the interaction network.
Notes:
- When I say pduA this refers to the specific sequence, not pduA family, which means there would eb a complex for each pduA from each organism not only 1 pduABJKN.
- If two papers independently assemble PduABJKN, this is one biological complex record with multiple observations/comments.



# Problem to address:


## Protein and complex:
Relationship: many to many.

A protein can:
- belong to zero, one, or many complexes
- directly interact with other proteins
- directly interact with itself
- interact with proteins that are not in the same complex;
- interact with proteins from a completely different BMC system, such as a Pdu protein interacting with EutS.

<b> The protein exists independently of any particular complex. This says nothing by itself about which proteins physically touch.

For example: pduA is a component of complex 1,2,3 and 4. This says nothing by itself about which proteins physically touch.

## IS essential
A particular complex may not assembly without an specific protein, while another protein may not be required and its lack will just lead to a different complex. what if stoiciometry varies based on this? Have stoiociometry as a variable instead of an absolute number?

## Stoichiometry
The stoichiometry of an specific protein in that particular complex.

# PPI (and interaction, merged)
https://journals.plos.org/ploscompbiol/article?id=10.1371/journal.pcbi.1004067#sec003 
A PPI means: <b> Protein X directly interacts with Protein Y. The unique unordered pair of proteins whose direct interaction status has been reported.

This is fundamentally different from merely belonging to the same complex.

For example: in complex 1 (pduABJKN) pduA may interact with B and N, but that does not mean K and N interact directly. 

- PPI in this database is considered bidirectional (e.g.: A-K and K-A interactions are the same). Any additional biolgical meaning is recorded in interaction. 
- Self interaction is allowed (e.g.: pduA can interact with pduA). Shell proteins self-association/oligomeric assembly can be part of shell formation.
- PPI does not depend on complex membership. So PPI must be able to exist without being attached to a complex.(E.g.: pduA interaction with pduU, but pduU is not in complex 1,2 and 4)
- One interaction can occur in multiple complexes (e.g.: pduA-K interaction relevant for complex 1, 2 and 3)
- We may know the complex but not the interaction network. So lack of PPI records does not mean these proteins do not interact, it can mean the pairwise interaction information is unknown.


- Interaction ifnormation may be incomplete, for example we can know two proteins interact wihtout knowing how. Absence of interaction should ot invalid ppi. 

- Repeated evidence for the same PPI: for example, Paper 1 experimentally shows A ↔ K, Paper 2 predicts A ↔ K, and Paper 3 investigates it again.This will be added as an A–K PPI with an evolving description
## Interaction type:
Add interaction_status to distinguish between: INTERACTION, NO_INTERACTION_DETECTED, if empty is unknonw. In interaction description the rest of the information can be added (for example whether is predicted, experimental, structural evidence...). Important to not use NO_INTERACTION, this should not be allowed in dictionary as we experiments are conditional. 

## Interaction descirption:
It can contain information such as:
- Experimentally demonstrated interaction...
- Predicted based on...
- Interaction observed under...
- Suggested interaction...
- Evidence currently limited...
- Structural modelling suggests...
- Further study reported...

## PPI_complex: REMOVE!!!!!!!!!!!

SQL is specifically designed to combine information across tables using joins, so this type of query is feasible.
SELECT DISTINCT ppi.*
FROM ppi

JOIN protein_complex pc1
    ON pc1.prot_id = ppi.prot_id_1

JOIN protein_complex pc2
    ON pc2.prot_id = ppi.prot_id_2

WHERE pc1.complex_id = :complex_id
  AND pc2.complex_id = :complex_id;

However, if protein-complex is removed the meaning of interaction changes:
Without ppi_complex, these two proteins are known to interact, AND both happen to be members of this complex.

It does not necessarily prove: This interaction was experimentally demonstrated within this particular complex.
<div>

<b> Summary of interaction: PPIs will be represented as unique, undirected protein pairs; self-interaction is permitted, absence of a pair means unknown, and an explicit NO_INTERACTION_DETECTED state distinguishes experimentally reported negative results from missing knowledge.
